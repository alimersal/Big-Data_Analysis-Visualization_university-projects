Purpose: This notebook analyzes global population data stored in a MongoDB database and visualizes population trends from **1980 to 2050** using interactive maps and charts.


| # | Library | Role |
|---|---------|------|
| 1 | `pymongo` | Connect to MongoDB & retrieve data |

## Cell 1: Interactive Marker Map
Connects to MongoDB, loads country data into a DataFrame, then renders a Folium world map.  
Each country marker shows a popup with population figures for: 1980, 2000, 2010, 2022, 2023, 2030, and 2050.



```python
# Import necessary libraries
import folium
from folium.plugins import MarkerCluster
import pandas as pd
import pymongo

# Connect to MongoDB
client = pymongo.MongoClient('mongodb://localhost:27017')
db = client['countries-table']
collection = db['countries']

# Load data into a pandas DataFrame
data = pd.DataFrame(list(collection.find()))

# Drop duplicates from DataFrame
data.drop_duplicates(inplace=True)

# Create a map centered at the mean of latitude and longitude from the dataset
m = folium.Map(location=[data['latitude'].mean(), data['longitude'].mean()], zoom_start=2)

# Customize the Marker
marker_cluster = MarkerCluster().add_to(m)

# Adding data to the map
for row in data.itertuples():
    popup_content = f'''
    <b>Country:</b> {row.country}<br>
    <b>Rank:</b> {row.rank}<br>
    <b>Population 1980:</b> {row.pop1980}<br>
    <b>Population 2000:</b> {row.pop2000}<br>
    <b>Population 2010:</b> {row.pop2010}<br>
    <b>Population 2022:</b> {row.pop2022}<br>
    <b>Population 2023:</b> {row.pop2023}<br>
    <b>Population 2030:</b> {row.pop2030}<br>
    <b>Population 2050:</b> {row.pop2050}<br>
    '''

    folium.Marker(location=[row.latitude, row.longitude],
                  popup=folium.Popup(popup_content, max_width=150, min_width=100),
                  icon=folium.Icon(color='black',icon='cloud')).add_to(marker_cluster)
# Show the map
m
```


## Cell 2: Choropleth World Map (2023 Population)
Fetches data from MongoDB, converts population columns to float, then draws a Plotly choropleth map.  
Color intensity reflects each country's population in 2023 using the `Portland` color scale.



```python
# Import necessary libraries
import pandas as pd
import plotly.express as px
from pymongo import MongoClient

# Connect to MongoDB
# Establish a connection
client = MongoClient('mongodb://localhost:27017')
# Select the database
db = client['countries-table']
# Select the collection
collection = db['countries']


# Edit the range for the color scale
color_scale_range = [0,155000000]  # Adjust the range as needed

# Fetch the data from MongoDB to pandas DataFrame
df = pd.DataFrame(list(collection.find()))

# Convert data types as needed
df['pop1980'] = df['pop1980'].astype(float)
df['pop2000'] = df['pop2000'].astype(float)
df['pop2010'] = df['pop2010'].astype(float)
df['pop2022'] = df['pop2022'].astype(float)
df['pop2023'] = df['pop2023'].astype(float)
df['pop2030'] = df['pop2030'].astype(float)
df['pop2050'] = df['pop2050'].astype(float)
# Create a choropleth map for population in the year 2023
fig = px.choropleth(
    df,
    locations="country",
    locationmode="country names",
    color="pop2023",
    hover_name="country",
    projection="natural earth",
    title="World Population in 2023",
    template="plotly",
    color_continuous_scale="Portland",
    range_color=color_scale_range,  # Set the range for the color scale
)

# Adjust the size of the figure
fig.update_layout(
    autosize=False,
    width=1000,  # Adjust the width as needed
    height=450,  # Adjust the height as needed
)

fig.show()
```


## Cell 3: Grouped Bar Chart (Population Over the Years)
Builds a grouped bar chart with Plotly Express comparing population across all countries  
for years: 1980, 2000, 2010, 2022, 2023, 2030, 2050. Includes a range slider and rotated x-axis labels.




```python
import plotly.express as px
import pandas as pd

# Assuming data is your DataFrame
fig2 = px.bar(
    df,
    x="country",
    y=["pop1980", "pop2000", "pop2010", "pop2022", "pop2023", "pop2030", "pop2050"],
    title="Population Over the Years",
    labels={"value": "Population", "variable": "Year"},
    template="plotly",
)

# Adjust the size of the figure and rotate x-axis labels
fig2.update_layout(
    height=500,  # Set the height in pixels
    width=900,  # Set the width in pixels
    xaxis_tickangle=-45,  # Rotate x-axis labels for better readability
)

# Set bar mode to 'group' to ensure all countries are visible
fig2.update_layout(barmode='group')

# Add rangeslider and rangeselector for zooming
fig2.update_layout(
    xaxis=dict(
        rangeslider=dict(visible=True),
        type="category",  # Set x-axis type to category
    ),
)

fig2.show()
```


## Cell 4: Scatter Plot (Population Over the Years)
Displays a scatter plot of population data points per country across multiple years (1980–2050).  
A range slider on the x-axis enables interactive zooming for detailed country comparison.



```python
# Assuming data is your DataFrame
fig2 = px.scatter(
    df,
    x="country",
    y=["pop1980", "pop2000", "pop2010", "pop2022", "pop2023", "pop2030", "pop2050"],
    title="Population Over the Years",
    labels={"value": "Population", "variable": "Year"},
    template="plotly",
)

# Adjust the size of the figure and rotate x-axis labels
fig2.update_layout(
    height=500,  # Set the height in pixels
    width=900,  # Set the width in pixels
    xaxis_tickangle=-45,  # Rotate x-axis labels for better readability
)

# Add rangeslider and rangeselector for zooming
fig2.update_layout(
    xaxis=dict(
        rangeslider=dict(visible=True),
        type="category",  # Set x-axis type to category
    ),
)

fig2.show()
```
