```python
# Import the necessary libraries
import pymongo
import pandas as pd
import matplotlib.pyplot as plt
from pymongo import MongoClient

# Connect to MongoDB
try:
    client = MongoClient('mongodb://localhost:27017/')
    db = client['Eartquakes-1990-2023']
    collection = db['prediction']
    print("Connected to MongoDB")
except pymongo.errors.ServerSelectionTimeoutError as err:
    print(f"Error connecting to MongoDB: {err}")
    # Print the entire traceback for more details
    import traceback
    traceback.print_exc()

# Load the data from MongoDB to a DataFrame
try:
    data = pd.DataFrame(list(collection.find()))
    # Continue with your data analysis or processing here
    print("Data loaded successfully")
except Exception as e:
    print(f"Error loading data: {e}")
    # Print the entire traceback for more details
    import traceback
    traceback.print_exc()
```
